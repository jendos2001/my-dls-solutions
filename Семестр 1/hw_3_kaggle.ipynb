Этап 3. GridSearchCV

```python
class featureExtractor(BaseEstimator, TransformerMixin):

    def fit(self, X, y=None):
        return self

    def transform(self, data):
        data = data.copy()
        data = data.replace({'TotalSpent': {' ': np.nan}})
        data = data.astype({'TotalSpent': np.float64})
        data = data.drop(columns=['HasPhoneService', 'HasInternetService'])
        data['CountServices'] = (data[services] == 'Yes').sum(axis=1)
        data = data.drop(columns=services)
        return data


data_grid_search = pd.read_csv('/content/train.csv')

num_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        ("num" ,num_transformer, new_num_cols),
        ("cat", OneHotEncoder(handle_unknown='ignore', sparse_output=False), new_cat_cols),
    ]
)

pipeline = Pipeline(
    [
        ("feature_extractor", featureExtractor()),
        ("encoder", preprocessor),
        ("model", LogisticRegression()),
    ]
)
```

```python
y_grid_search = data_grid_search[target_col]
X_grid_search = data_grid_search.drop(target_col, axis=1)
```

```python
params = [
    {
        'model__penalty': ['l1'],
        'model__C': [10 ** n for n in range(2, -4, -1)],
        'model__solver': ['liblinear', 'saga']
    },
    {
        'model__penalty': ['l2'],
        'model__C': [10 ** n for n in range(2, -4, -1)],
        'model__solver': ['lbfgs', 'liblinear', 'newton-cg', 'newton-cholesky', 'sag', 'saga']
    },
    {
        'model__penalty': ['elasticnet'],
        'model__C': [10 ** n for n in range(2, -4, -1)],
        'model__solver': ['saga']
    }
]
```

```python
grid = GridSearchCV(pipeline, params, refit=True, cv=5, scoring='roc_auc', n_jobs=1, verbose=3)
grid.fit(X_grid_search, y_grid_search)
```
